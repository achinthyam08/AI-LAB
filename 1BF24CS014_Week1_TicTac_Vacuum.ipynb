```python
board = [" "] * 9

def print_board():
    print(f"{board[0]} | {board[1]} | {board[2]}")
    print("--+---+--")
    print(f"{board[3]} | {board[4]} | {board[5]}")
    print("--+---+--")
    print(f"{board[6]} | {board[7]} | {board[8]}")

def check_winner():
    winning_positions = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] and board[a] != " ":
            return board[a]

    if " " not in board:
        return "Draw"

    return None

player = "X"

while True:
    print_board()

    position = int(input(f"Player {player}, enter position (1-9): ")) - 1

    if board[position] != " ":
        print("Position already occupied!")
        continue

    board[position] = player

    winner = check_winner()

    if winner:
        print_board()

        if winner == "Draw":
            print("Game Draw!")
        else:
            print("Player", winner, "wins!")

        break

    player = "O" if player == "X" else "X"
```

```python
def simple_reflex_agent(location, rooms):
    if rooms[location] == "Dirty":
        return "Suck"
    else:
        return "Move"


# Many rooms
rooms = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Dirty",
    "D": "Dirty",
    "E": "Clean"
}

location = "A"

room_names = list(rooms.keys())

while True:
    print("Rooms:", rooms)
    print("Current Location:", location)

    action = simple_reflex_agent(location, rooms)
    print("Action:", action)

    if action == "Suck":
        rooms[location] = "Clean"

    elif action == "Move":
        current_index = room_names.index(location)
        next_index = (current_index + 1) % len(room_names)
        location = room_names[next_index]

    print()

    # Stop when all rooms are clean
    if all(status == "Clean" for status in rooms.values()):
        print("All rooms are clean!")
        break
```

```python
def goal_based_agent(rooms, location):
    # Goal: all rooms must be clean

    if rooms[location] == "Dirty":
        return "Suck"

    # Find a dirty room
    for room in rooms:
        if rooms[room] == "Dirty":
            return "Move to " + room

    return "Goal Achieved"


# Many rooms
rooms = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Dirty",
    "D": "Dirty",
    "E": "Clean",
    "F": "Dirty"
}

location = "A"

while True:
    print("Rooms:", rooms)
    print("Current Location:", location)

    action = goal_based_agent(rooms, location)
    print("Action:", action)

    if action == "Suck":
        rooms[location] = "Clean"

    elif action.startswith("Move to"):
        destination = action.split()[-1]
        location = destination

    elif action == "Goal Achieved":
        print("All rooms are clean!")
        break

    print()
```
