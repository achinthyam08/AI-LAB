```python

import random
import math

N = 8

# Calculate number of attacking queen pairs
def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Same row or same diagonal
            if board[i] == board[j]:
                conflicts += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Generate a neighboring state
def get_neighbor(board):
    new_board = board[:]

    column = random.randint(0, N - 1)
    new_row = random.randint(0, N - 1)

    while new_row == new_board[column]:
        new_row = random.randint(0, N - 1)

    new_board[column] = new_row

    return new_board


# Simulated Annealing algorithm
def simulated_annealing():
    current = [random.randint(0, N - 1) for _ in range(N)]
    current_cost = heuristic(current)

    temperature = 100.0
    cooling_rate = 0.99
    minimum_temperature = 0.001

    while current_cost != 0 and temperature > minimum_temperature:

        neighbor = get_neighbor(current)
        neighbor_cost = heuristic(neighbor)

        delta = neighbor_cost - current_cost

        # Accept better or equal states
        if delta <= 0:
            current = neighbor
            current_cost = neighbor_cost

        # Sometimes accept worse states
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = neighbor
                current_cost = neighbor_cost

        # Reduce temperature
        temperature *= cooling_rate

    return current, current_cost


# Display the chessboard
def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


solution, cost = simulated_annealing()

print("Final Board:")
display_board(solution)

print("Number of conflicts:", cost)

if cost == 0:
    print("Solution found!")
else:
    print("Solution not found. Try running again.")

```
