```python

import math

def alpha_beta(node, alpha, beta):
    # Leaf node
    if isinstance(node, int):
        return node

    node_type = node[0]
    children = node[1:]

    if node_type == "MAX":
        value = -math.inf

        for i, child in enumerate(children):
            value = max(value, alpha_beta(child, alpha, beta))
            alpha = max(alpha, value)

            if alpha >= beta:
                print("MAX: Pruned remaining children:", children[i + 1:])
                break

        return value

    else:  # MIN node
        value = math.inf

        for i, child in enumerate(children):
            value = min(value, alpha_beta(child, alpha, beta))
            beta = min(beta, value)

            if alpha >= beta:
                print("MIN: Pruned remaining children:", children[i + 1:])
                break

        return value


# Example game tree
tree = [
    "MAX",
    ["MIN", 3, 5],
    ["MIN", 2, 9],
    ["MIN", 4, 6]
]

result = alpha_beta(tree, -math.inf, math.inf)

print("Final value of MAX node:", result)

```
