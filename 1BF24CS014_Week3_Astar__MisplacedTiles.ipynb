```python
import heapq

START = "283164705"
GOAL = "123804765"


# -------------------------------------------------
# Display the puzzle
# -------------------------------------------------
def print_board(state):
    print(f"{state[0]} {state[1]} {state[2]}")
    print(f"{state[3]} {state[4]} {state[5]}")
    print(f"{state[6]} {state[7]} {state[8]}")


# -------------------------------------------------
# Generate ALL valid children
# -------------------------------------------------
def get_neighbors(state):
    neighbors = []

    blank = state.index("0")
    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, direction in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether move is valid
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            temp = list(state)

            # Swap blank and tile
            temp[blank], temp[new_position] = \
                temp[new_position], temp[blank]

            child = ''.join(temp)

            neighbors.append((child, direction))

    return neighbors


# -------------------------------------------------
# Heuristic 1: Misplaced Tiles
# -------------------------------------------------
def misplaced_tiles(state, goal):

    h = 0

    for i in range(9):

        # Do not count blank
        if state[i] != '0' and state[i] != goal[i]:
            h += 1

    return h


# -------------------------------------------------
# Heuristic 2: Manhattan Distance
# -------------------------------------------------
def manhattan_distance(state, goal):

    h = 0

    for tile in "12345678":

        current = state.index(tile)
        target = goal.index(tile)

        current_row = current // 3
        current_col = current % 3

        target_row = target // 3
        target_col = target % 3

        h += abs(current_row - target_row)
        h += abs(current_col - target_col)

    return h


# -------------------------------------------------
# A* with ALL expansions displayed
# -------------------------------------------------
def a_star(start, goal, heuristic):

    # Priority queue:
    # (f, counter, g, state, path)
    open_list = []

    counter = 0

    g_start = 0
    h_start = heuristic(start, goal)
    f_start = g_start + h_start

    heapq.heappush(
        open_list,
        (f_start, counter, g_start, start, [start])
    )

    visited = set()

    current_depth = -1

    while open_list:

        f, _, g, state, path = heapq.heappop(open_list)

        if state in visited:
            continue

        visited.add(state)

        # Print depth heading
        if g != current_depth:

            current_depth = g

            print("\n")
            print("=" * 60)
            print("DEPTH =", g)
            print("=" * 60)

        h = heuristic(state, goal)

        print("\nEXPANDED STATE")
        print("-" * 40)

        print_board(state)

        print(f"g = {g}")
        print(f"h = {h}")
        print(f"f = {f}")

        # Goal
        if state == goal:

            print("\nGOAL REACHED!")
            print("Solution depth =", g)

            return path

        # Generate ALL children
        children = get_neighbors(state)

        print("\nALL VALID EXPANSIONS:")

        for child, direction in children:

            if child in visited:
                continue

            new_g = g + 1
            new_h = heuristic(child, goal)
            new_f = new_g + new_h

            counter += 1

            print("\nMove:", direction)

            print_board(child)

            print(
                f"g = {new_g}, "
                f"h = {new_h}, "
                f"f = {new_f}"
            )

            heapq.heappush(
                open_list,
                (
                    new_f,
                    counter,
                    new_g,
                    child,
                    path + [child]
                )
            )

    return None


# =================================================
# CASE 1
# MISPLACED TILES
# =================================================

print("\n")
print("#" * 60)
print("A* SEARCH - MISPLACED TILES")
print("#" * 60)

path1 = a_star(
    START,
    GOAL,
    misplaced_tiles
)


# =================================================
# CASE 2
# MANHATTAN DISTANCE
# =================================================

# print("\n\n")
# print("#" * 60)
# print("A* SEARCH - MANHATTAN DISTANCE")
# print("#" * 60)

# path2 = a_star(
#     START,
#     GOAL,
#     manhattan_distance
# )
```
