```python
from itertools import product

def implies(p, q):
    return not p or q

def tt_entails(kb_function, query_function, symbols):
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb = kb_function(model)
        query = query_function(model)

        # Print every truth assignment
        print(model, "KB =", kb, "Query =", query)

        # Counterexample: KB is true but query is false
        if kb and not query:
            return False

    return True

# EXAMPLE 1: KB entails Q
# KB = (P -> Q) AND P
def kb1(m):
    return implies(m['P'], m['Q']) and m['P']

def query1(m):
    return m['Q']

print("EXAMPLE 1: KB entails Q")
result1 = tt_entails(kb1, query1, ['P', 'Q'])
print("Entails:", result1)

# EXAMPLE 2: KB does not entail Q
# KB = P -> Q
def kb2(m):
    return implies(m['P'], m['Q'])

def query2(m):
    return m['Q']

print("\nEXAMPLE 2: KB does not entail Q")
result2 = tt_entails(kb2, query2, ['P', 'Q'])
print("Entails:", result2)
```
