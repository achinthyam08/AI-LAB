```python
from itertools import product

def implies(p, q):
    return not p or q

def print_truth_table(kb_func, query_func, symbols):
    # Header of the table
    header = f"| {' | '.join(symbols)} | P->Q | Q->R |  P   |  KB  | Query |"
    separator = "-" * len(header)
    print(header)
    print(separator)

    entails = True
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))

        p, q, r = model['P'], model['Q'], model['R']
        cond1 = implies(p, q)
        cond2 = implies(q, r)
        cond3 = p
        kb = cond1 and cond2 and cond3

        # Initialize Query value representation
        query_str = "-"
        if kb:
            query_val = query_func(model)
            query_str = str(query_val)
            if not query_val:
                entails = False

        # Print the row with aligned columns
        print(f"| {str(p):<4} | {str(q):<4} | {str(r):<4} | {str(cond1):<4} | {str(cond2):<4} | {str(cond3):<4} | {str(kb):<4} | {query_str:<5} |")

    print(separator)
    print("Entails:", entails)



# KB = (P -> Q) AND (Q -> R) AND P
def kb1(m):
    return implies(m['P'], m['Q']) and implies(m['Q'], m['R']) and m['P']

def query1(m):
    return m['R']

def query2(m):
    return not m['P']

def query3(m):
    return implies(m['Q'], m['P'])

print("EXAMPLE 1: Query is R")
print_truth_table(kb1, query1, ['P', 'Q', 'R'])

print("\nEXAMPLE 2: Query is not P")
print_truth_table(kb1, query2, ['P', 'Q', 'R'])

print("\nEXAMPLE 3: Query is Q -> P")
print_truth_table(kb1, query3, ['P', 'Q', 'R'])
```
