```python
def calculate_cost(state):
    """Count pairs of queens attacking diagonally."""
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbors(state):
    """Generate neighbors by swapping every pair of queens."""
    n = len(state)
    neighbors = []

    for i in range(n):
        for j in range(i + 1, n):
            new_state = list(state)

            # Swap queens at columns i and j
            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append(((i, j), tuple(new_state),
                              calculate_cost(new_state)))

    return neighbors


def hill_climbing(initial_state):
    current = tuple(initial_state)
    step = 0

    while True:
        current_cost = calculate_cost(current)

        print("\nStep:", step)
        print("Current state:", current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("\nGoal configuration reached!")
            print("Final state:", current)
            return

        neighbors = generate_neighbors(current)

        print("\nNeighbor states and costs:")
        for pair, state, cost in neighbors:
            print("Swap", pair,
                  "->", state,
                  "| Cost =", cost)

        # Minimum cost; ties are resolved by smallest pair (i, j)
        best = min(neighbors, key=lambda item: (item[2], item[0]))

        pair, next_state, next_cost = best

        if next_cost >= current_cost:
            print("\nNo better neighbor found.")
            print("Hill climbing terminated.")
            return

        print("\nSelected swap:", pair)
        print("Next state:", next_state)
        print("Next cost:", next_cost)

        current = next_state
        step += 1


# Initial configuration from the question
initial_state = (1,2,3,0)

hill_climbing(initial_state)
```
