```python

facts = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]



rules = [
    ([("Missile", "x")],
     ("Weapon", "x")),


    ([("Enemy", "x", "America")],
     ("Hostile", "x")),


    ([("Missile", "x"), ("Owns", "A", "x")],
     ("Sells", "Robert", "x", "A")),


    ([("American", "p"),
      ("Weapon", "q"),
      ("Sells", "p", "q", "r"),
      ("Hostile", "r")],
     ("Criminal", "p"))
]


def unify(pattern, fact, theta=None):

    if theta is None:
        theta = {}

    if pattern[0] != fact[0]:
        return None

    if len(pattern) != len(fact):
        return None

    for p, f in zip(pattern[1:], fact[1:]):

        if p.islower() and len(p) == 1:

            if p in theta:
                if theta[p] != f:
                    return None
            else:
                theta[p] = f

        elif p != f:
            return None

    return theta



def substitute(expression, theta):

    return tuple(theta.get(x, x) for x in expression)


def forward_chaining(facts, rules, query):

    kb = list(facts)

    print("INITIAL KNOWLEDGE BASE:")
    for fact in kb:
        print(fact)

    print("\nFORWARD CHAINING STARTED\n")

    while True:

        new_facts = []

        for premises, conclusion in rules:


            substitutions = [{}]


            for premise in premises:

                updated_substitutions = []

                for theta in substitutions:

                    for fact in kb:

                        new_theta = unify(
                            premise,
                            fact,
                            theta.copy()
                        )

                        if new_theta is not None:
                            updated_substitutions.append(new_theta)

                substitutions = updated_substitutions


            for theta in substitutions:

                new_fact = substitute(conclusion, theta)

                if new_fact not in kb and new_fact not in new_facts:

                    new_facts.append(new_fact)

                    print("New fact derived:", new_fact)


                    if unify(query, new_fact) is not None:
                        print("\nQUERY PROVED!")
                        return True


        if not new_facts:
            break


        kb.extend(new_facts)

    print("\nQUERY NOT PROVED")
    return False



query = ("Criminal", "Robert")


result = forward_chaining(facts, rules, query)

print("\nFINAL RESULT:", result)

```
