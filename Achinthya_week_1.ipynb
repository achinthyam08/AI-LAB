```python
board = [" "] * 9

def print_board():
    print(f"{board[0]} | {board[1]} | {board[2]}")
    print("--+---+--")
    print(f"{board[3]} | {board[4]} | {board[5]}")
    print("--+---+--")
    print(f"{board[6]} | {board[7]} | {board[8]}")

def check_winner():
    winning_positions = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] and board[a] != " ":
            return board[a]

    if " " not in board:
        return "Draw"

    return None

player = "X"

while True:
    print_board()

    position = int(input(f"Player {player}, enter position (1-9): ")) - 1

    if board[position] != " ":
        print("Position already occupied!")
        continue

    board[position] = player

    winner = check_winner()

    if winner:
        print_board()

        if winner == "Draw":
            print("Game Draw!")
        else:
            print("Player", winner, "wins!")

        break

    player = "O" if player == "X" else "X"
```

```python
def simple_reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    else:
        return "Move Left"


# Environment
location = "A"
status = "Dirty"

for i in range(5):
    print("Location:", location)
    print("Status:", status)

    action = simple_reflex_agent(location, status)
    print("Action:", action)
    print()

    if action == "Suck":
        status = "Clean"
    elif action == "Move Right":
        location = "B"
        status = "Dirty"
    elif action == "Move Left":
        location = "A"
        status = "Dirty"
```

```python
def goal_based_agent(environment, location):
    # Goal: both rooms should be clean

    if environment["A"] == "Dirty" and location == "A":
        return "Suck"

    if environment["B"] == "Dirty" and location == "B":
        return "Suck"

    if environment["A"] == "Dirty":
        return "Move to A"

    if environment["B"] == "Dirty":
        return "Move to B"

    return "Goal Achieved"


# Environment
environment = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"

while True:
    print("Environment:", environment)
    print("Current Location:", location)

    action = goal_based_agent(environment, location)
    print("Action:", action)
    print()

    if action == "Suck":
        environment[location] = "Clean"

    elif action == "Move to A":
        location = "A"

    elif action == "Move to B":
        location = "B"

    elif action == "Goal Achieved":
        print("Both rooms are clean!")
        break
```
